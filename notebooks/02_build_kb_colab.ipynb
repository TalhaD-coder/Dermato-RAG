## 2. Makaleleri Yükle
**DİKKAT:** GitHub'da veri klasörü olmadığı için, PubMed'den çektiğimiz makaleleri buraya yüklemeniz gerekiyor.

Bilgisayarınızdaki `data/knowledge_base/` klasörünü ZIP'leyip sol taraftaki dosya menüsünden yükleyin ve aşağıdaki kodu çalıştırın:

```python
import os
import zipfile
from pathlib import Path

# Eğer bilgisayarınızdan "knowledge_base.zip" yüklediyseniz:
zip_path = "knowledge_base.zip"
target_dir = Path("data")

if os.path.exists(zip_path):
    print("Zipten çıkarılıyor...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(target_dir)
    print("Makaleler başarıyla yüklendi!")
else:
    print(f"LÜTFEN DİKKAT: Sol taraftan {zip_path} dosyasını yükleyin!")
    # Sadece klasörleri oluştur
    Path("data/knowledge_base/raw_docs").mkdir(parents=True, exist_ok=True)
```

## 4. Sonuçları Bilgisayarına Geri İndir
Bilgi tabanı ChromaDB'ye işlendi (`data/embeddings/chromadb/`).
Bunu ziple ve bilgisayarına indir.

```python
import shutil
from google.colab import files

if os.path.exists("data/embeddings/chromadb"):
    print("Veritabanı zıpleniyor...")
    shutil.make_archive("chromadb_export", 'zip', "data/embeddings/chromadb")
    print("İndirme başlıyor...")
    files.download("chromadb_export.zip")
else:
    print("ChromaDB klasörü bulunamadı!")
```
